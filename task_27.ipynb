```python
import pandas as pd

data = {
    'Name': ['Uma', 'Sirisha', 'Akshitha', 'Navya'],
    'Age': [19, 20, 19, 21],
    'Marks': [85, 90, 78, 95],
    'City': ['Guntur', 'Ongole', 'Vijayawada', 'Tenali']
}

df = pd.DataFrame(data)

df.to_excel('students.xlsx', index=False)

print("Excel file created successfully!")
```

```python
import pandas as pd

df = pd.read_excel('students.xlsx')

# Filter students with marks above 80
top_students = df[df['Marks'] > 80]

# Sort marks in descending order
sorted_data = top_students.sort_values(
    by='Marks', ascending=False
)

# Export processed data
sorted_data.to_excel(
    'processed_students.xlsx',
    index=False
)

print("Processed Excel file saved successfully!")
print(sorted_data)
```

```python
import pandas as pd

df = pd.read_excel('students.xlsx')

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing marks
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Add a new column
df['Result'] = df['Marks'].apply(
    lambda x: 'Pass' if x >= 40 else 'Fail'
)

print("\nCleaned Data:")
print(df)
```
